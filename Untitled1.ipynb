```python
import kagglehub
import os
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix

# 1. Downloading dataset from Kaggle
path = kagglehub.dataset_download("tongpython/cat-and-dog")
train_dir = os.path.join(path, 'training_set', 'training_set')
test_dir = os.path.join(path, 'test_set', 'test_set')

# 2. Loading images and resizing(data preprocessing!-)
train_data = tf.keras.utils.image_dataset_from_directory(
    train_dir, image_size=(160, 160), batch_size=32
)
test_data = tf.keras.utils.image_dataset_from_directory(
    test_dir, image_size=(160, 160), batch_size=32, shuffle=False
)

# 3. Load pre trained cnn MobileNetV2 (a part of neural network!)
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(160, 160, 3), include_top=False, weights='imagenet'
)
base_model.trainable = False  # Freeze the pre-trained weights

# 4. Build a simple model
model = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"), # Basic data augmentation
    tf.keras.layers.Rescaling(1./255),        # Normalization beign done
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(), # Flatten the features
    tf.keras.layers.Dense(1, activation='sigmoid') # activation func!
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)#loss func!!

# 5. Train for just 3 epochs (hyperparameter!)
history = model.fit(train_data, validation_data=test_data, epochs=3)

# 6.  plot for Accuracy
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Test Accuracy')
plt.title('Model Accuracy')
plt.legend()
plt.show()

# 7.  plot for Loss
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Test Loss')
plt.title('Model Loss')
plt.legend()
plt.show()

# 8. Get predictions for evaluation
y_true = []
y_pred = []

for images, labels in test_data:
    preds = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    # If probability is > 0.5 guess Dog (1) else  Cat (0)
    y_pred.extend([1 if p > 0.5 else 0 for p in preds])

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=['Cat', 'Dog']))

print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred))
```

```python
import numpy as np
from google.colab import files
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Ask the user to upload file
print("Please upload a picture of a Cat or Dog...")
uploaded = files.upload()

# 2. Process the uploaded image
for filename in uploaded.keys():
    print("\nProcessing", filename, "...")

    # set hyperparameters
    img = tf.keras.utils.load_img(filename, target_size=(160, 160))

    # Show image
    plt.imshow(img)
    plt.axis('off')
    plt.show()

    # Convert the image to numbers
    img_array = tf.keras.utils.img_to_array(img)

    # Add an extra dimension (the model expects a "batch" of images, even if it's just a batch of 1)
    img_array = tf.expand_dims(img_array, 0)

    # 3. Make the prediction!
    prediction = model.predict(img_array)
    probability = prediction[0][0]

    # 1 is dog 0 is cat
    if probability > 0.5:
        dog_confidence = round(probability * 100, 2)
        print("Prediction: DOG! (Confidence:", dog_confidence, "%)")
    else:

        cat_confidence = round((1 - probability) * 100, 2)
        print("Prediction: CAT! (Confidence:", cat_confidence, "%)")
```
